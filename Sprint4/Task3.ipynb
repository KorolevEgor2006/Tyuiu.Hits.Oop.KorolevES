Создать базовый класс Order в C#, который будет представлять информацию о 
заказах товаров или услуг. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма.

Требования к базовому классу Order: 
• Атрибуты: ID заказа (OrderId), Дата создания (CreationDate), Сумма заказа 
(TotalAmount). 
• Методы: 
o CalculateTotal(): метод для расчета общей суммы заказа. 
o AddItem(Item item): метод для добавления элемента в заказ. 
o RemoveItem(Item item): метод для удаления элемента из заказа. 

Требования к производным классам: 
1. ОнлайнЗаказ (OnlineOrder): Должен содержать дополнительные атрибуты, 
такие как Email клиента (CustomerEmail). Метод AddItem() должен быть 
переопределен для добавления информации о способе доставки при 
добавлении элемента.

2. ФизическийЗаказ (PhysicalOrder): Должен содержать дополнительные

3. СпециализированныйЗаказ (SpecializedOrder) (если требуется третий класс):

```
using System;

public interface IOnlineOrder
{
    public void SetBankCard(string _bankcard);
    public void SendNotification();
    public void SendNotification(string customMessage);
}
public interface ISpecializedOrder
{
    public void AddPoints(int points); 
    public void SetBonus(int points);
    public void SetBonus(double discount);
}
public interface IOrder
{
    public void AddItemsIsOtherOrder(Order other);
    public int GetCountItem(); 
    public string GetInfo();
}
public interface IPhysicalOrder
{
    public void ChangeAddress(string newAddress);
    public void ChangeAddress(string newAddress,string _aboutAddress);
    public void SetComment(string comment);
}
public abstract class Order:IOrder
{
    private int _orderId;
    private DateTime _creationDate;
    private double _totalAmount;
    private double _minimalAmount;
    private double _maximalAmount;
    private double _mediumAmount;
    
    public static int TotalOrder{get;private set;}

    public int OrderId 
    {
        get{return _orderId;}
        set{_orderId=value;}
    }
    public DateTime CreationDate
    {
        get{return _creationDate;}
        protected set{_creationDate=value;}
    }
    public double TotalAmount
    {
        get{return _totalAmount;}
        protected set{_totalAmount=value;}
    }
    public double MinimalAmount
    {
        get{return _minimalAmount;}
        protected set{_minimalAmount=value;}
    }
    public double MaximalAmount
    {
        get{return _maximalAmount;}
        protected set{_maximalAmount=value;}
    }
    public double MediumAmount
    {
        get{return _mediumAmount;}
        protected set{_mediumAmount=value;}
    }
    protected List<Item> items; 
    public Order(int id,DateTime date)
    {
        OrderId=id;
        CreationDate=date;
        TotalAmount=0;
        TotalOrder++;
        items= new List<Item>();
    }
    public virtual void AddItem(Item item)
    {
        items.Add(item);
        Console.WriteLine($"В заказ {OrderId} - добавлено {item.ItemName} стоимостью {item.ItemCost}");
        CalculateTotal();
        ProcessingItems();
    }
    public virtual void RemoveItem(Item item)
    {
        items.Remove(item);
        Console.WriteLine($"Из заказа {OrderId} - удален {item.ItemName} стоимостью {item.ItemCost}");
        CalculateTotal();
        ProcessingItems();
    }
    public virtual void CalculateTotal()
    {
        double cost=0;
        foreach(Item item in items)
        {
            cost+=item.ItemCost;
        }
        TotalAmount=cost;
    }
    void IOrder.AddItemsIsOtherOrder(Order other)
    {
        List<Item> otherItems=other.items;
        Console.WriteLine($"Добавление в заказ {OrderId} из заказа {other.OrderId}:");
        foreach(Item item in otherItems)
        {
            AddItem(item);
        }
        Console.WriteLine($"Из заказа {other.OrderId} в заказ {OrderId} было добавлено {otherItems.Count} позиции.");
    }
    public virtual void InfoOrder()
    {
        Console.WriteLine($"Заказ {OrderId} создан {CreationDate:yy.MM.dd.HH.mm.ss}. Итого:{TotalAmount}");
        Console.WriteLine($"Минимальная стоимость товара: {MinimalAmount}. Максимальная стоимость товара: {MaximalAmount}. Средняя стоимость товара: {MediumAmount}.");
    }
    public virtual void ProcessingItems()
    {
        MinimalAmount=items.OrderBy(item => item.ItemCost).FirstOrDefault().ItemCost;
        MaximalAmount=items.OrderByDescending(item => item.ItemCost).FirstOrDefault().ItemCost;
        MediumAmount=Math.Round(items.Count > 0 ? items.Average(item => item.ItemCost) : 0,2);
    }
    int IOrder.GetCountItem()
    {
        return items.Count;
    }
    string IOrder.GetInfo()
    {
        return $"Заказ {OrderId} создан {CreationDate:yy.MM.dd.HH.mm.ss}. Общая стоимость:{TotalAmount}";
    }
}

public class Item
{
    private int itemId;
    private string itemName;
    private double itemCost;

    public int ItemId
    {
        get{return itemId;}
        set{itemId=value;}
    }
    public string ItemName
    {
        get{return itemName;}
        set{itemName=value;}
    }
    public double ItemCost
    {
        get{return itemCost;}
        set{itemCost=value;}
    }
    public Item(int id, string name, double cost)
    {
        ItemId=id;
        ItemName=name;
        ItemCost=cost;
    }
}

public class ItemOnline:Item
{
    private string delivery;

    public string Delivery{
        get{return delivery;}
        set{delivery=value;}
    }
    public ItemOnline(int id,string name,double cost,string delivery):base(id,name,cost)
    {
        Delivery=delivery;
    }
}

public class OnlineOrder: Order,IOnlineOrder
{
    private string customerEmail;
    private string bankcard;

    public string CustomerEmail
    {
        get{return customerEmail;}
        set{
            if(value.Contains("@"))customerEmail=value;
            else throw new ArgumentException($"Некорректный адрес электронной почты.{value}");
        }
    }
    public string BankCard
    {
        get{return bankcard;}
        set{
            if((value.Length==22)&&(value.Contains(":") && value.Contains("-")))
            {
                if((value[16]==':')&&(value[19]=='-'))bankcard=value;
                else throw new ArgumentException($"Некорректный номер банковской карты.{value}");
            }
            else throw new ArgumentException($"Некорректная длина номера банковской карты.{value}");
        }
    }

    public OnlineOrder(int id,DateTime date,string email):base(id,date)
    {
        CustomerEmail=email;
    }
    public override void AddItem(Item item)
    {
        base.AddItem(item);
        if(item is ItemOnline itemOnline)
        {
            Console.WriteLine($"{CustomerEmail}:В заказе {OrderId} - {itemOnline.ItemName} будет доставлен {itemOnline.Delivery}");
        }
    }
    public override void InfoOrder()
    {
        Console.WriteLine($"Заказ {OrderId} пользователем с электронной почтой {CustomerEmail} создан {CreationDate:yy.MM.dd.HH.mm.ss}. Итого:{TotalAmount}");
        Console.WriteLine($"Минимальная стоимость товара: {MinimalAmount}. Максимальная стоимость товара: {MaximalAmount}. Средняя стоимость товара: {MediumAmount}.");
    }
    void IOnlineOrder.SendNotification()
    {
        Console.WriteLine($"Уведомление на электронную почту {CustomerEmail} о доставке заказа {OrderId}");
    }
    void IOnlineOrder.SendNotification(string customMessage)
    {
        Console.WriteLine($"Уведомление на электронную почту {CustomerEmail} о заказе {OrderId}: {customMessage}");
    }
    void IOnlineOrder.SetBankCard(string _bankcard)
    {
        BankCard=_bankcard;
        ((IOnlineOrder)this).SendNotification($"Привязана банковская карта - {BankCard}");
    }
}

public class ConfirmedOnlineOrder: OnlineOrder
{
    private string ipAddress;
    private string phone;
    public string IPAddress{
        get{return ipAddress;}
        protected set{ipAddress=value;}
    }
    public string Phone{
        get{return phone;}
        protected set{phone=value;}
    }
    public ConfirmedOnlineOrder(int id,DateTime date,string email,string _ipAddress,string _phone):base(id,date,email)
    {
        IPAddress=_ipAddress;
        Phone=_phone;
    }
    public override void InfoOrder()
    {
        Console.WriteLine($"Заказ {OrderId} ,подтвержденный пользователем с номером телефона {Phone} и с электронной почтой {CustomerEmail}, создан на устройстве с ip-адрессом {IPAddress} - {CreationDate:yy.MM.dd.HH.mm.ss}. Итого:{TotalAmount}");
        Console.WriteLine($"Минимальная стоимость товара: {MinimalAmount}. Максимальная стоимость товара: {MaximalAmount}. Средняя стоимость товара: {MediumAmount}.");
    }
}

public class PhysicalOrder:Order,IPhysicalOrder
{
    private string deliveryAddress;
    private string aboutAddress;
    private string commentInOrder;
    public string DeliveryAddress
    {
        get{return deliveryAddress;}
        set{deliveryAddress=value;}
    }
    public string AboutAddress
    {
        get{return aboutAddress;}
        set{aboutAddress=value;}
    }
    public string CommentInOrder
    {
        get{return commentInOrder;}
        set{commentInOrder=value;}
    }
    public PhysicalOrder(int id,DateTime date,string address):base(id,date)
    {
        DeliveryAddress=address;
        AboutAddress="";
        CommentInOrder="";
    }
    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);
        Console.WriteLine($"{item.ItemName} будет возвращен из {DeliveryAddress}");
    }
    public override void InfoOrder()
    {
        Console.WriteLine($"Заказ {OrderId} создан {CreationDate:yy.MM.dd.HH.mm.ss} будет доставлен по адресу: {DeliveryAddress}. Итого:{TotalAmount}");
        Console.WriteLine($"Минимальная стоимость товара: {MinimalAmount}. Максимальная стоимость товара: {MaximalAmount}. Средняя стоимость товара: {MediumAmount}.");
        Console.WriteLine($"Комментарий к заказу:{CommentInOrder}");
    }
    void IPhysicalOrder.ChangeAddress(string newAddress)
    {
        DeliveryAddress=newAddress;
        Console.WriteLine($"Новый адрес доставки:{DeliveryAddress}");
    }
    void IPhysicalOrder.ChangeAddress(string newAddress,string _aboutAddress)
    {
        DeliveryAddress=newAddress;
        AboutAddress=_aboutAddress;
        Console.WriteLine($"Новый адрес доставки:{DeliveryAddress}. О новом адресе: {AboutAddress}");
    }
    void IPhysicalOrder.SetComment(string comment)
    {
        CommentInOrder=comment;
        Console.WriteLine($"К заказу {OrderId} установлен комментарий:{CommentInOrder}");
    }
    
}

public class SpecialConditions
{
    private double discount;
    private int bonuses;
    public double Discount
    {
        get{return discount;}
        set{discount=value;}
    }
    public int Bonuses
    {
        get{return bonuses;}
        set{bonuses=value;}
    }
    public SpecialConditions(double discount, int bonuses)
    {
        Discount=discount;
        Bonuses=bonuses;
    }
}

public class SpecializedOrder:Order, ISpecializedOrder
{
    private SpecialConditions specialConditions;
    private int addedPoints;
    private bool isGift;
    public SpecialConditions _specialConditions
    {
        get{return specialConditions;}
        set{specialConditions=value;}
    }
    public int AddedPoints
    {
        get{return addedPoints;}
        set{addedPoints=value;}
    }
    public bool IsGift
    {
        get{return isGift;}
        set{isGift=value;}
    }
    public SpecializedOrder(int id,DateTime date,SpecialConditions specialConditions,bool _isGift):base(id,date)
    {
        _specialConditions=specialConditions;
        AddedPoints=0;
        IsGift=_isGift;
    }
    public override void CalculateTotal()
    {
        double cost=0;
        foreach(Item item in items)
        {
            cost+=item.ItemCost;
        }
        if(IsGift)cost+=30;
        cost-=_specialConditions.Bonuses;
        if(cost<0)cost=0;
        cost*=1-(_specialConditions.Discount/100);
        TotalAmount=cost;
    }
    public override void InfoOrder()
    {
        Console.WriteLine($"Заказ {OrderId} создан {CreationDate:yy.MM.dd.HH.mm.ss}.Списано {_specialConditions.Bonuses} бонусов. Скидка {_specialConditions.Discount}%. Итого:{TotalAmount}");
        Console.WriteLine($"Всего начислено бонусов:{AddedPoints}");
        if(IsGift)Console.WriteLine("Будет оформлен, как праздничный заказ.");
        Console.WriteLine($"Минимальная стоимость товара: {MinimalAmount}. Максимальная стоимость товара: {MaximalAmount}. Средняя стоимость товара: {MediumAmount}.");
    }
    void ISpecializedOrder.AddPoints(int points)
    {
        _specialConditions.Bonuses+=points;
        AddedPoints+=points;
        Console.WriteLine($"Добавлено {points} баллов");
    }
    void ISpecializedOrder.SetBonus(int points)
    {
        AddedPoints+=(points - _specialConditions.Bonuses);
        _specialConditions.Bonuses=points;
        Console.WriteLine($"Установленно {points} баллов");
    }
    void ISpecializedOrder.SetBonus(double discount)
    {
        _specialConditions.Discount=discount;
        Console.WriteLine($"Установленна скидка {discount}");
    }
}

public class OrderCollection<T> where T : Order 
{ 
    private List<T> orders = new List<T>();
    public void AddOrder(T order)
    {
        orders.Add(order);
    }

    public void RemoveOrder(T order)
    {
        orders.Remove(order);
    }
    public void DisplayInfoOrders()
    {
        foreach(var order in orders)
        {
            order.InfoOrder();
        }
    }
    public T GetOrder(int id)
    {
        return orders.FirstOrDefault(o=>o.OrderId==id);
    }
}
public class OrderService
{
    private readonly IOrder _order;
    public OrderService(IOrder order)
    {
        _order=order;
    }

    public void DisplayShortInfo()
    {
        Console.WriteLine(_order.GetInfo());
        Console.WriteLine($"Всего предметов в заказе:{_order.GetCountItem()}");
    }

    public void AddItemOtherOrder(Order order)
    {
        _order.AddItemsIsOtherOrder(order);
    }
}

OrderCollection<Order> Orders = new OrderCollection<Order>();

OnlineOrder onlineOrder= new OnlineOrder(1,new DateTime(2025,9,8,8,0,46),"exemple@mail.ru");
onlineOrder.AddItem(new ItemOnline(1,"Тарелка",120.39,"курьером"));
onlineOrder.AddItem(new ItemOnline(2,"Кружка",99.89,"курьером"));
Orders.AddOrder(onlineOrder);
Console.WriteLine();
PhysicalOrder physicalOrder = new PhysicalOrder(2,new DateTime(2025,9,7,22,0,46),"Россия,г.Магнитошахтинск,ул. Голубей 1000");
physicalOrder.AddItem(new Item(3,"Миксер",490.48));
Item returnedItem =new Item(3,"Миксер",490.48);
physicalOrder.AddItem(returnedItem);
physicalOrder.AddItem(new Item(4,"Кастрюля",290.99));
physicalOrder.AddItem(new Item(7,"Зеркало",499.99));
Orders.AddOrder(physicalOrder);
physicalOrder.RemoveItem(returnedItem);
((IPhysicalOrder)physicalOrder).ChangeAddress("Россия,г.Магнитошахтинск,ул. Голубей 100,25");
((IPhysicalOrder)physicalOrder).ChangeAddress("Россия,г.Магнитошахтинск,ул. Голубей 100,25", "Подъезд 3. Нумерация подъездов начинается справа.");
((IPhysicalOrder)physicalOrder).SetComment("Заказ хрупкий");
IOrder physical = physicalOrder;
OrderService physicalService = new OrderService(physical);
physicalService.DisplayShortInfo();
Console.WriteLine();
SpecializedOrder specializedOrder = new SpecializedOrder(3,new DateTime(2025,9,7,11,0,46),new SpecialConditions(50,50),true);
Orders.AddOrder(specializedOrder);
specializedOrder.AddItem(new Item(5,"Хлеб",60));
returnedItem =new Item(5,"Хлеб",60);
specializedOrder.AddItem(returnedItem);
specializedOrder.RemoveItem(returnedItem);
specializedOrder.AddItem(new Item(6,"Молоко",40));
((ISpecializedOrder)specializedOrder).SetBonus(20.0);
((ISpecializedOrder)specializedOrder).SetBonus(0);
((ISpecializedOrder)specializedOrder).AddPoints(100);
Console.WriteLine();
ConfirmedOnlineOrder confirmedOnlineOrder = new ConfirmedOnlineOrder(4,new DateTime(2025,9,8,18,0,46),"exemple@mail.ru","127.0.0.1","+7-000-000-00-00");
confirmedOnlineOrder.AddItem(new ItemOnline(4,"Кастрюля",290.99,"курьером"));
Orders.AddOrder(confirmedOnlineOrder);
IOrder confirmOnline = confirmedOnlineOrder;
OrderService confirderOnlineService = new OrderService(confirmOnline);
confirderOnlineService.DisplayShortInfo();
confirderOnlineService.AddItemOtherOrder(onlineOrder);
confirderOnlineService.DisplayShortInfo();
Console.WriteLine();
IOnlineOrder receivedOrder = Orders.GetOrder(1) as IOnlineOrder;
receivedOrder.SendNotification();
receivedOrder.SetBankCard("0020932010239283:26-09");
Console.WriteLine();
Console.WriteLine($"Всего заказов {Order.TotalOrder}");
Orders.DisplayInfoOrders();
```
