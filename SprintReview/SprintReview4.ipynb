```polyglot-notebook
public interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List<ISupplier> GetSuppliers();
    decimal CalculateTotalExpenses(); 
}

public interface ISupplier
{
    string GetName();
    decimal GetCost();
    string SupplierType{get; set;}
}

public interface IEmployee
{
    string GetFullName();
    decimal GetSalary();
    string Position{get;set;}
}

public class Enterprise:IEnterprise
{
    private List<ISupplier> Suppliers;
    private List<IEmployee> Employees;
    public Enterprise()
    {
        Suppliers = new List<ISupplier>();
        Employees = new List<IEmployee>();
    }
    public void AddSupplier(ISupplier supplier)
    {
        Suppliers.Add(supplier);
    }
    public void AddEmployee(IEmployee employee)
    {
        Employees.Add(employee);
    }
    public List<ISupplier> GetSuppliers()
    {
        return Suppliers;
    }
    public List<IEmployee> GetEmployees()
    {
        return Employees;
    }

    public decimal CalculateTotalExpenses()
    {
        decimal sum=0;
        foreach(var s in Suppliers)
        {
            sum+=s.GetCost();
        }
        foreach(var e in Employees)
        {
            sum+=e.GetSalary();
        }
        return sum;
    }
}

public class Supplier:ISupplier
{
    private string supplierType;
    private string Name;
    private decimal Cost;
    public string SupplierType
    {
        get{return supplierType;}
        set{supplierType=value;}
    }

    public Supplier(string name,decimal cost, string _supplierType)
    {
        Name=name;
        Cost=cost;
        SupplierType=_supplierType;
    }
    public string GetName()
    {
        return Name;
    }

    public decimal GetCost()
    {
        return Cost;
    }
 
}

public class Employee:IEmployee
{
    private string name;
    private string position;
    private decimal salary;

    public Employee(string Name, string _Position, decimal Salary)
    {
        name= Name;
        Position=_Position;
        salary= Salary;
    }

    public string Position
    {
        get{return position;}
        set{position = value;}
    }

    public string GetFullName()
    {
        return name;
    }

    public decimal GetSalary()
    {
        return salary;
    }

}

Enterprise enterprise = new Enterprise();
Supplier supplier1 = new Supplier("Логистический центр",100000,"Логистика");
Supplier supplier2 = new Supplier("Поставщик офисной мебели",50000,"Офисное оборудование");
enterprise.AddSupplier(supplier1);
enterprise.AddSupplier(supplier2);
Employee employee1 = new Employee("Иван Иванов","Менеджер",70000);
Employee employee2 = new Employee("Анна Петрова","Разработчик",95000);
enterprise.AddEmployee(employee1);
enterprise.AddEmployee(employee2);
List<ISupplier> suppliers = enterprise.GetSuppliers();
List<IEmployee> employees = enterprise.GetEmployees();
Console.WriteLine("Список поставщиков:");
foreach(var s in suppliers)
{
    Console.WriteLine($"{s.GetName()} - Стоимость услуг: {s.GetCost()} - Тип: {s.SupplierType}");
}
Console.WriteLine("Список сотрудников:");
foreach(var e in employees)
{
    Console.WriteLine($"{e.GetFullName()} - Должность:{e.Position} - Зарплата: {e.GetSalary()}");
}
Console.WriteLine($"Общие расходы: {enterprise.CalculateTotalExpenses()}");
```
