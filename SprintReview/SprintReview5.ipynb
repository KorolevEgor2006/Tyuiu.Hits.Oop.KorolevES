```polyglot-notebook
using System;
using System.Linq;

public delegate void DelegateOrder();

public class Company
{
    private string Name;
    private List<Employee> Employees;
    private List<Order> Orders;

    public string name{get{return Name;} set{Name=value;}}
    public Company(string _name)
    {
        Employees = new List<Employee>();
        Orders= new List<Order>();
        name=_name;
    }
    public void AddEmployee(Employee employee)
    {
        Employees.Add(employee);
    }
    public void AddOrder(Order order)
    {
        Orders.Add(order);
    }
    public void AddAssignedEmployee(int _OrderId,Employee employee)
    {
        var orderAdd =  (Order)Orders.Where(order => order.orderId==_OrderId);
        orderAdd.assignedEmployee=employee;
        orderAdd.SetStatus("В работе");
        employee.ordersProcessed++;
    }
    public void SearchStatusOrder(string status)
    {
        Console.WriteLine($"Заказы со статусом {status}");
        foreach(var order in Orders )
        {
            
            if(order.OrderStatus==status)Console.WriteLine($"Заказ {order.orderId}");
        }
    }
    public void SearchOrderCustomer(Customer customer)
    {
        Console.WriteLine($"Заказы заказчика {customer.firstName} {customer.lastName} :");
        foreach(var order in Orders )
        {
            
            if(order.customerOrder==customer)Console.WriteLine($"Заказ {order.orderId}");
        }
    }
}

public class Employee
{
    private string FirstName;
    private string LastName;
    private string Position;
    private int OrdersProcessed;

    public string firstName{get{return FirstName;} set{FirstName=value;}}
    public string lastName{get{return LastName;} set{LastName=value;}}
    public string position{get{return Position;} set{Position=value;}}
    public int ordersProcessed{get {return OrdersProcessed;} set {OrdersProcessed=value;}}
    
    public Employee(string _firstname, string _lastName, string _position)
    {
        firstName=_firstname;
        lastName=_lastName;
        position=_position;
        OrdersProcessed=0;
    }
}

public class Customer
{
    private string FirstName;
    private string LastName;
    private string PhoneNumber;

    public string firstName{get{return FirstName;} set{FirstName=value;}}
    public string lastName{get{return LastName;} set{LastName=value;}}
    public string phoneNumber{get{return PhoneNumber;} set{PhoneNumber=value;}}
    public Customer(string _firstname, string _lastName, string _phoneNumber)
    {
        firstName=_firstname;
        lastName=_lastName;
        phoneNumber=_phoneNumber;
    }
}
public class Order
{   
    private static int OrdersCount=0;
    private int OrderId;
    private string Description;
    private Employee AssignedEmployee;
    //private enum OrderStatus;
    private string orderStatus;
    private DateTime CreatedAt;
    private Customer CustomerOrder;

    public delegate void DelegateOrderChangeStatus(string status, string statusOld);
    public event DelegateOrderChangeStatus OrderChangeStatus;

    public int orderId{get{return OrderId;} set{OrderId=value;}}
    public string description{get{return Description;} set{Description=value;}}
    public Employee assignedEmployee{get{return AssignedEmployee;} set{AssignedEmployee=value;}}
    public Customer customerOrder{get{return CustomerOrder;} set{CustomerOrder=value;}}
    public string OrderStatus {get{return orderStatus;} private set{orderStatus=value;}}

    public Order(string _description, Customer customer)
    {
        orderId=OrdersCount;
        customerOrder= customer;
        OrdersCount++;
        OrderStatus= "Создан";
        CreatedAt=DateTime.Now;
        description=_description;
    }
    public void SetStatus(string status)
    {
        OrderChangeStatus?.Invoke(status,orderStatus); 
        orderStatus = status;
    }
}

Company company = new Company("Компания 1");
Employee e1 = new Employee("Иван", "Иванов", "Менеджер");
Customer c1 = new Customer("Илья", "Петров", "8 000 000 00 00");
Customer c2 = new Customer("Андрей", "Петров", "8 001 000 00 00");
Order o1 = new Order("Описание",c1);
Order o2 = new Order("Описание",c2);
company.AddOrder(o1);
company.AddOrder(o2);
company.AddEmployee(e1);
company.SearchOrderCustomer(c2);
//company.AddAssignedEmployee(0,e1);
o2.OrderChangeStatus+=(string status, string statusOld)=>Console.WriteLine($"Статус заказа {o2.orderId} изменился с {statusOld} на {status}");
o2.SetStatus("Завершенно");
company.SearchStatusOrder("Создан");
```
