```polyglot-notebook
public class Employee
{
    private string name;
    private string position;
    private float salary;

    public string Name
    {
        get{return name;}
        set{name=value;}
    }
    public string Position
    {
        get{return position;}
        set{position=value;}
    }
    public float Salary
    {
        get{return salary;}
        set{salary=value;}
    }
    
    public Employee(string _name, string _position, float _salary)
    {
        Name=_name;
        Position=_position;
        Salary=_salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Сотрудник {Name}, Должность: {Position}, Зарплата: {Salary}");
    }
}

public interface IReportable
{
    public string GenerateReport();
}

public interface ITeamLeader
{
    public string[] GetTeamMembers();
}

public class Manager: Employee
{
    private string departament;

    public string Departament
    {
        get{return departament;}
        set{departament=value;}
    }
    public Manager(string _name, string _position, float _salary,string _departament):base(_name, _position,_salary)
    {
        Departament=_departament;
    }
    public override void PrintInfo()
    {
        Console.WriteLine($"Менеджер отдела {Departament} - {Name}, Должность: {Position}, Зарплата: {Salary}");
    }
}

public class ProjectManager: Manager,IReportable,ITeamLeader
{
    private string projectName;
    private string[] teamMembers;
    public string ProjectName
    {
        get{return projectName;}
        set{projectName=value;}
    }
    public string[] TeamMembers
    {
        get{return teamMembers;}
        set{teamMembers=value;}
    }
    public ProjectManager(string _name, string _position, float _salary,string _departament, string _projectName)
    :base(_name, _position,_salary,_departament)
    {
        ProjectName=_projectName;
    }
    public override void PrintInfo()
    {
        Console.WriteLine($"Менеджер отдела {Departament} - {Name}, Должность: {Position}, Зарплата: {Salary}. Отвечает за проект {ProjectName}");
    }

    public string GenerateReport()
    {
        return $"Отчет по проекту {ProjectName} менеджера {Name}из отдела {Departament}.";
    }

    public string[] GetTeamMembers()
    {
        return TeamMembers;
    }
}
Employee employee = new Employee("Иванов Иван Иванович","Сотрудник",50000);
employee.PrintInfo();
Manager manager =  new Manager("Петров Петр Петрович","Менеджер",80000,"Электротехники");
manager.PrintInfo();
ProjectManager projectManager = new ProjectManager("Петров Андрей Андреевич","Проектный менеджер",96000,"Электротехники","Экономия стоимости производства");
projectManager.TeamMembers=new string[]{"Сотрудник 1","Сотрудник 2"};
projectManager.PrintInfo();
Console.WriteLine(projectManager.GenerateReport());
Console.WriteLine($"Количество сотрудников:{projectManager.GetTeamMembers().Length}");


```
