```polyglot-notebook
public class Component
{
    private string name;
    private float price;
    
    public string Name
    {
        get{return name;}
        set{name=value;}
    }
    public float Price
    {
        get{return price;}
        set{price=value;}
    }

    public Component(string _name,float _price)
    {
        Name=_name;
        Price=_price;
    }
    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент {Name} стоимостью{ToString(Price)}");
    }
    public string ToString(float s)
    {
        return s.ToString();
    }

} 
public interface IInstallable
{
    public void Install();
}

public class CPU:Component,IInstallable
{
    private int countCore;
    private float gercCore;
    private bool install;
    public bool _install
    {
        get{return install;}
        set{install=value;}
    }
    public int CountCore
    {
        get{return countCore;}
        set{countCore=value;}
    }
    public float GercCore
    {
        get{return gercCore;}
        set{gercCore=value;}
    }
    public CPU(string _name,float _price,int _countCore,float _gercCore):base(_name,_price)
    {
        CountCore=_countCore;
        GercCore=_gercCore;
    }
    public void Install()
    {
        _install=true;
        Console.WriteLine($"CPU {Name} установлен");
    }

    public override void GetInfo()
    {
        Console.WriteLine($"CPU {Name} стоимостью{ToString(Price)} рублей c {countCore} ядрами частатой {GercCore} Гц");
    }
    public static bool operator ==(CPU left, CPU right)
    {
        if(left.Name==right.Name&&left.Price==right.Price&&left.CountCore==right.CountCore&&left.GercCore==right.GercCore&&left._install==right._install)
        {
            return true;
        }
        return false;
    }
    public static bool operator !=(CPU left, CPU right)
    {
        if(left.Name==right.Name&&left.Price==right.Price&&left.CountCore==right.CountCore&&left.GercCore==right.GercCore&&left._install==right._install)
        {
            return false;
        }
        return true;
    }

}
public class RAM:Component,IInstallable
{
    private int member;
    private bool install;
    public int Member
    {
        get{return member;}
        set{member=value;}
    }
    public bool _install
    {
        get{return install;}
        set{install=value;}
    }
    public RAM(string _name,float _price,int _member):base(_name,_price)
    {
        Member=_member;
    }
    public void Install()
    {
        _install=true;
        Console.WriteLine($"RAM {Name} установлен");
    }
    public override void GetInfo()
    {
        Console.WriteLine($"RAM {Name} стоимостью: {ToString(Price)} рублей и c оперативной памятью:{Member} Mb");
    }
    public static bool operator ==(RAM left, RAM right)
    {
        if(left.Name==right.Name&&left.Price==right.Price&&left.Member==right.Member&&left._install==right._install)
        {
            return true;
        }
        return false;
    }
    public static bool operator !=(RAM left, RAM right)
    {
        if(left.Name==right.Name&&left.Price==right.Price&&left.Member==right.Member&&left._install==right._install)
        {
            return false;
        }
        return true;
    }
}
public class GPU:Component,IInstallable
{
    private int member;
    private bool install;
    public int Member
    {
        get{return member;}
        set{member=value;}
    }
    public bool _install
    {
        get{return install;}
        set{install=value;}
    }
    public GPU(string _name,float _price,int _member):base(_name,_price)
    {
        Member=_member;
    }
    public void Install()
    {
        _install=true;
        Console.WriteLine($"GPU {Name} установлен");
    }
    public override void GetInfo()
    {
        Console.WriteLine($"GPU {Name} стоимостью: {ToString(Price)} рублей и c видеопамятью:{Member} Mb");
    }
    public static bool operator ==(GPU left, GPU right)
    {
        if(left.Name==right.Name&&left.Price==right.Price&&left.Member==right.Member&&left._install==right._install)
        {
            return true;
        }
        return false;
    }
    public static bool operator !=(GPU left, GPU right)
    {
        if(left.Name==right.Name&&left.Price==right.Price&&left.Member==right.Member&&left._install==right._install)
        {
            return false;
        }
        return true;
    }
}
public class HDD:Component,IInstallable
{
    private int member;
    private bool install;
    public int Member
    {
        get{return member;}
        set{member=value;}
    }
    public bool _install
    {
        get{return install;}
        set{install=value;}
    }
    public HDD(string _name,float _price,int _member):base(_name,_price)
    {
        Member=_member;
        _install=false;
    }
    public void Install()
    {
        _install=true;
        Console.WriteLine($"HDD {Name} установлен");
    }
    public override void GetInfo()
    {
        Console.WriteLine($"HDD {Name} стоимостью: {ToString(Price)} рублей и c памятью:{Member} Gb");
    }
    public static bool operator ==(HDD left, HDD right)
    {
        if(left.Name==right.Name&&left.Price==right.Price&&left.Member==right.Member&&left._install==right._install)
        {
            return true;
        }
        return false;
    }
    public static bool operator !=(HDD left, HDD right)
    {
        if(left.Name==right.Name&&left.Price==right.Price&&left.Member==right.Member&&left._install==right._install)
        {
            return false;
        }
        return true;
    }
}
public class Computer<T> where T:Component
{
    private List<T> components = new List<T>();
    private string name;
    
    public string Name
    {
        get{return name;}
        set{name=value;}
    }
    public void AddComponent(T component)
    {
        components.Add(component);
    }
    public void RemoveComponent(T component)
    {
        components.Remove(component);        
    }
    public float GetTotalPrice()
    {
        float res=0;
        foreach(var component in components)
        {
            res+=component.Price;
        }
        return res;
    }
}
Computer<Component> Computer1 = new Computer<Component>();
Computer1.Name="Компьютер 1";
CPU cpu1 = new CPU("Intel i7",20000,8,2.4f);
CPU cpu2 = new CPU("Intel i7",20000,8,2.4f);
RAM ram1= new RAM("King",6000,8224);
GPU gpu1 = new GPU("Видеокарта",40000,4000);
HDD hdd1 = new HDD("Seagate",10000,1028);
Computer1.AddComponent(cpu1);
cpu1.GetInfo();
cpu1.Install();
Computer1.AddComponent(gpu1);
gpu1.Install();
gpu1.GetInfo();
Computer1.AddComponent(ram1);
ram1.Install();
ram1.GetInfo();
Computer1.AddComponent(hdd1);
hdd1.GetInfo();
hdd1.Install();
Computer1.RemoveComponent(gpu1);
Console.WriteLine($"Общая стоимость компонентов в {Computer1.Name}:{Computer1.GetTotalPrice()} рублей");
Console.WriteLine(cpu1!=cpu2);
```
